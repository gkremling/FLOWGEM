```python
import importlib
import subprocess

def install_if_missing(package, import_name=None):
    name = import_name or package
    if importlib.util.find_spec(name) is None:
        subprocess.run(["pip", "install", package], check=True)
```

## Set parameters and load existing results if available

```python
import pyreadr
from uci_datasets import Dataset
import pandas as pd
import numpy as np
import os

os.makedirs("./data/uci", exist_ok=True)

datasets = ["wine", "energy", "parkinsons", "stock", "pumadyn32nm",
            "housing", "forest", "bike"]

for name in datasets:
    ds = Dataset(name)
    x_train, y_train, x_test, y_test = ds.get_split(split=0)
    
    # Combine train and test for full dataset
    x = np.concatenate([x_train, x_test], axis=0)
    y = np.concatenate([y_train, y_test], axis=0)
    
    df = pd.DataFrame(x.astype(np.float64))
    df["target"] = y.astype(np.float64)
    
    pyreadr.write_rds(f"./data/uci/{name}.RDS", df)
    print(f"Saved {name}: n={len(df)}, d={df.shape[1]}")
```
