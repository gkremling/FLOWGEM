```python
import pyreadr
import torch
import pandas as pd
import numpy as np
from wgf_minimal import impute_wgf_python

datasets = ["fat", "windspeed", "airfoil_self_noise", "enb", "scm1d", "yeast", 
            "stock", "energy", "wine", "gas", "solar", "forest", 
            "housing", "pumadyn32nm", "parkinsons"]

```

```python


for name in datasets:
    try:
        np.random.seed(42)
        torch.manual_seed(42)
        
        data = f'mar.0.1.1.{name}.RDS'
        print(f"\nProcessing: {data}")
        
        result = pyreadr.read_r(f'./results/amputed/{data}')
        X_obs = result[None]
        
        Xhats = impute_wgf_python(X_obs)
        
        pyreadr.write_rds(f'./results/imputed/wgf.{data}', pd.DataFrame(Xhats))
        print(f"Saved: wgf.{data}")
        
    except Exception as e:
        print(f"Failed: {name} - {e}")
```
